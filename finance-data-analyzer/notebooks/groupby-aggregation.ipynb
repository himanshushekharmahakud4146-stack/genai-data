# GroupBy and Aggregation Analysis

## Objective

Analyze employee data by grouping records into meaningful categories
and calculating summary statistics using Pandas.

## Topics Covered

- GroupBy
- Group size and count
- Mean, sum, minimum and maximum
- Multiple aggregations with `agg()`
- Multi-column grouping
- Sorting grouped results

## Dataset

kaggle_pandas.csv


```python
import pandas as pd
reviews = pd.read_csv("../data/raw/kaggle_pandas.csv", index_col=0)
pd.set_option("display.max_rows", 5)
```

```python
reviews.head()
```

```python
reviews.groupby("country").count()
```

```python
reviews.groupby("points").points.count()
```

```python
reviews.groupby("points").price.min()
```

```python
reviews.groupby("points").price.max()
```

```python
reviews.groupby('winery').apply(lambda df: df.title.iloc[0])
```

```python
reviews.groupby(['country','province']).apply (lambda df: df.loc[df.points.idxmax()])
```

```python
reviews.groupby('country').price.agg([len, min, max])
```

```python
countries_reviewed = reviews.groupby(['country', 'province']).description.agg([len])
countries_reviewed
```

```python
countries_reviewed.reset_index()
```

```python
reviews.head()
```

```python
# Q.1
reviews_written = reviews.groupby('taster_twitter_handle').taster_twitter_handle.count()
reviews_written
```

```python
# Q.2
best_rating_per_price = reviews.groupby('price')['points'].max().sort_index()
best_rating_per_price
```

```python
# Q.3   
price_extremes = reviews.groupby('variety').price.agg([min, max])
price_extremes
```

```python
# Q.4
sorted_varieties = price_extremes.sort_values(by = ['min','max'],ascending = False)
sorted_varieties
```

```python
#Q.5
reviewer_mean_ratings = reviews.groupby('taster_name').points.mean()
reviewer_mean_ratings
```

```python
# Q.6
country_variety_counts = reviews.groupby(['country','variety']).size().sort_values(ascending = False)
country_variety_counts
```
