## 2. Selecting Relevant Columns

For analysis, only relevant employee, department, income, and satisfaction fields are selected.

Selecting specific columns makes the analysis easier to read and keeps the workflow focused on the business questions.

```python
high_income = df[
    df["MonthlyIncome"] > 5000
]

high_income.head()
```

```python
experienced_employees = df[
    (df["YearsAtCompany"] > 5) &
    (df["MonthlyIncome"] < 5000)
]

experienced_employees.head()
```

```python
experienced_employees.sort_values(
    "MonthlyIncome",ascending=False
).head(10)
```

## 3. Filtering and Sorting

Conditional filtering is used to identify employees matching specific business criteria.

The analysis also uses sorting to rank employees based on monthly income.

```python
avg_income_by_department = (
    df.groupby("Department")["MonthlyIncome"]
      .mean()
      .sort_values(ascending=False)
)

avg_income_by_department
```

```python
employee_count_by_department = (
    df.groupby("Department")
      .size()
      .sort_values(ascending=False)
)

employee_count_by_department
```

```python
department_income_summary = (
    df.groupby("Department")["MonthlyIncome"]
      .agg(["mean", "min", "max", "count"])
)

department_income_summary
```
