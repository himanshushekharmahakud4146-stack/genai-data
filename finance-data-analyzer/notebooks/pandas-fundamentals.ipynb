# Subsetting Data

```python
df.columns
```

```python
# subsetting columns
df = df[['title', 'description', 'publishTime', 'duration_seconds', 'viewCount',
       'likeCount']]
```

```python
df.head()
```

```python
#subsetting using loc
df_subset1 = df.loc[df['viewCount'] > 1000000]
```

```python
df_subset2 = df.query('viewCount > 1000000')
```

```python
df = df.loc[~df['viewCount'].isna()]
```

# Casting dtypes

```python
df.dtypes
```

```python
df['viewCount'] = df['viewCount'].astype('int')
```

```python
df['viewCount'] = df['viewCount'].astype('int')
```

```python
df.dtypes
```

```python
df['publishTime'] = pd.to_datetime(df['publishTime'])
```

```python
df['likeCount'].astype('str')
```

```python
df['likeCount'] = pd.to_numeric(
    df['likeCount'],
    errors='coerce'
)
```

```python
df.dtypes
```
